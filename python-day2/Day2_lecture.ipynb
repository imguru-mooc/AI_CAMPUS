### ▶ 예제 1-1 · 어제의 예고를 코드로
`b = a` 는 복사가 아닙니다. 실행 전에 마지막 줄 출력을 예측해 보세요.

```python
a = [1, 2, 3]
b = a          # 이름표 하나 더
b.append(4)
print("a =", a)
print("a is b:", a is b)
```

### ▶ 예제 1-2 · 복사 3형제
`copy()` / `a[:]` / `list(a)` — 겉 리스트를 새로 만듭니다.

```python
a = [1, 2, 3]
c = a.copy()
c.append(99)
print(a, c, a is c)
```

### ▶ 예제 1-3 · 얕은 복사의 한계
안쪽 리스트는 여전히 공유됩니다. 🎬 02 "[Day 2] 중첩 얕은 복사" 와 함께 보세요.

```python
import copy
outer = [[10, 20], 3]
shallow = outer.copy()
deep = copy.deepcopy(outer)
shallow[0].append(30)
print("outer  :", outer)     # 안쪽이 함께 바뀜!
print("deep   :", deep)      # 완전 독립
```

### ▶ 예제 1-4 · sort() vs sorted()
🎬 D2-3 "⚠ a = a.sort() 함정" 프리셋과 함께.

```python
nums = [3, 1, 2]
print(sorted(nums), nums)   # 새 리스트, 원본 유지
print(nums.sort(), nums)    # None, 원본이 정렬됨
```

### ▶ 예제 1-5 · 튜플과 언패킹

### ▶ 예제 3-1 · 정의와 호출

```python
def adc_to_volt(raw, vref=3.3, bits=10):
    """ADC 원시값 → 전압"""
    return raw / (2**bits - 1) * vref

print(adc_to_volt(512))                 # 위치 인자 + 기본값
print(adc_to_volt(512, vref=5.0))       # 키워드 인자
print(round(adc_to_volt(2048, bits=12), 3))
```

### ▶ 예제 3-3 · 가변 인자와 다중 반환

```python
def stats(*values):
    return min(values), max(values), round(sum(values) / len(values), 2)

lo, hi, avg = stats(22.1, 23.4, 31.2)
print(lo, hi, avg)
```

### ▶ 예제 3-5 · 가변 기본값 함정
🎬 02 참조 모델의 원리가 그대로 적용됩니다.

```python
def bad_log(v, acc=[]):     # 기본 리스트는 정의 때 1회 생성
    acc.append(v)
    return acc
print(bad_log(1), bad_log(2))   # [1] [1, 2] — 공유·누적!

def safe_log(v, acc=None):
    if acc is None:
        acc = []
    acc.append(v)
    return acc
print(safe_log(1), safe_log(2)) # [1] [2]
```

### ▶ 예제 6-4 · raise 와 사용자 정의 예외

```python
class FrameError(Exception):
    pass

def parse_line(line):
    line = line.strip()
    if not (line.startswith("$") and line.endswith("*")):
        raise FrameError(f"손상 프레임: {line!r}")
    d = {}
    for f in line[1:-1].split(","):
        k, v = f.split("=")
        d[k] = v
    return (float(d["T"]), int(d["H"]), int(d["D"]))

print(parse_line("$T=25.3,H=60,D=120*"))
try:
    parse_line("broken")
except FrameError as e:
    print("FrameError:", e)
```
