### ▶ 예제 2-1 · 변수는 이름표

```python
a = 10
b = a          # 값 복사가 아니라 같은 객체에 이름표 하나 더
a = 20         # a 만 새 객체로 이동
print(a, b)
print(id(a) == id(b))
```

### ▶ 예제 2-2 · 기본 자료형과 `type()`

```python
adc_raw = 512            # int
vref = 3.3               # float
sensor_on = True         # bool
device_name = "ESP32"    # str
nothing = None           # 값 없음
for v in (adc_raw, vref, sensor_on, device_name, nothing):
    print(repr(v), type(v).__name__)
```

### ▶ 예제 2-3 · int 는 자릿수 제한이 없다 (C 의 int32 와 다름)

### ▶ 예제 2-4 · float 는 오차가 있다 (IEEE 754)

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(round(0.1 + 0.2, 2) == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
print(1e-3, 1.5e3, float("inf"))
```

### ▶ 예제 2-5 · 형변환 — 성공과 실패

```python
print(int("42"), float("3.14"), str(100) + "회", bool(3))
print(int(3.99), round(3.99), int(-3.99))    # int() 는 버림
print(int(True), float(False))
print(bool(0), bool(""), bool([]), bool(None))
print(bool("0"), bool("False"), bool(" "))   # 비어 있지 않으면 모두 True
```

```python
# 실패하는 변환 — 오류 메시지를 읽어 두세요
try:
    int("3.5")
except ValueError as e:
    print("ValueError:", e)
print(int(float("3.5")))   # 두 단계로
```

### ▶ 예제 2-6 · f-string 형식 지정 모음

```python
v = 1.6516; n = 4773; name = "ESP32"; big = 1234567
print(f"{v:.2f} | {v:.4f} | {v:8.1f}|")      # 소수점 자리, 폭
print(f"{n:>8} | {n:<8} | {n:^8} |")          # 오른쪽/왼쪽/가운데 정렬
print(f"{n:08d} | {n:,} | {big:,}")           # 0 채움, 천 단위 구분
print(f"{n:b} | {n:016b} | {n:o} | {n:x} | {n:#x} | {n:#06X}")   # 진법
print(f"{0.4567:.1%} | {name!r} | {name.upper()}")
```

### ▶ 예제 3-3 · 비트 연산자 기본

```python
a = 0b1100_1010
print(f"a        = {a:08b} ({a})")
print(f"a & 0x0F = {a & 0x0F:08b}")
print(f"a | 0x0F = {a | 0x0F:08b}")
print(f"a ^ 0xFF = {a ^ 0xFF:08b}")
print(f"a << 1   = {(a << 1) & 0xFF:08b}  (8비트로 자름)")
print(f"a >> 2   = {a >> 2:08b}")
print(f"~a       = {~a}  →  ~a & 0xFF = {~a & 0xFF:08b}")
```

### ▶ 예제 3-4 · 비트 마스크 4패턴 — GPIO 레지스터 시뮬레이션

```python
PORT = 0b0000_0000
LED  = 1 << 3          # 3번 핀
BTN  = 1 << 5          # 5번 핀

PORT |= LED;              print(f"설정  PORT |= LED     → {PORT:08b}")
PORT &= ~LED & 0xFF;      print(f"해제  PORT &= ~LED    → {PORT:08b}")
PORT ^= LED;              print(f"토글  PORT ^= LED     → {PORT:08b}")
PORT ^= LED;              print(f"토글  다시           → {PORT:08b}")
PORT = 0b0010_0000
print("읽기  PORT & BTN     →", "pressed" if PORT & BTN else "released")
```

### ▶ 예제 3-5 · 여러 핀을 한 번에

```python
PINS = (1 << 0) | (1 << 2) | (1 << 4) | (1 << 6)   # 짝수 핀 4개
print(f"{PINS:08b} = {PINS:#04x} = {PINS}")
PORT = 0
PORT |= PINS
print(f"{PORT:08b}")
PORT &= ~(1 << 2) & 0xFF                            # 2번만 끄기
print(f"{PORT:08b}")
```
